```python
# Convert anumpy array to a tensor
numpy_array = np.array([0.0,1.0,2.0,3.0,4.0])
new_tensor = torch.from_numpy(numpy_array)

print('the dtype of new tensor :',new_tensor.dtype)
print('the type of new tensor :',new_tensor.type())
```

```python
# Convert a tensor a numpy
back_to_numpy = new_tensor.numpy()

print('the dtype of new tensor :',back_to_numpy)
print('the dtype of new tensor :',back_to_numpy.dtype)
```

```python
# set all element in numpy array to zero
numpy_array[:] = 0
print('The new tensor point to numpy array:',new_tensor)
print('back to numpy array point to the tensor:',back_to_numpy)
```

```python
# Convert  panda series to a tensor
pandas_series = pd.Series([0.1,2,0.3,10.1])
new_tensor = torch.from_numpy(pandas_series.values)
print(new_tensor)
print(new_tensor.dtype)
print(new_tensor.type())
```

```python
selected_indexes = [3,4]
subset_tensor_sample = tensor_sample[selected_indexes]
print('The inital tensor sample', tensor_sample)
print('The subset of tensor sample with the values on index 3 and 4:', subset_tensor_sample)
```

### Tensor function

```python
# Craete a tensor 
math_tensor = torch.tensor([1.0,-1,1,-1])
print(math_tensor)
```

```python
# Calculate the mean for math_tensor
mean = math_tensor.mean()
print(mean)
```

```python
# Calculate the standard deviation for math_tensor
std_dev = math_tensor.std()
print(std_dev)
```

```python
# Min , Max
mm_tensor = torch.tensor([1,1,3,5,5])
print(mm_tensor)
```

```python
# Max , Min values
max_val = mm_tensor.max()
min_val = mm_tensor.min()
print(max_val)
print(min_val)
```

```python
# Method for calculating the sin result of each element in the tensor
pi_tensor = torch.tensor([0, np.pi/2, np.pi])
sin = torch.sin(pi_tensor)
print('The sin result of pi_tensor:', sin)
```

### Create Tensor by torch.linspace()

```python
# First try on using linspace to create tensor
len_5_tensor = torch.linspace(-2, 2, steps=5)
print('First try on linspace', len_5_tensor)
```

```python
# Assign steps with 9
len_9_tensor = torch.linspace(-2, 2, steps=9)
print(len_9_tensor)
```

```python
# Construct the tensor within 0 to 360 degree
pi_tensor = torch.linspace(0, 2*np.pi, 100)
sin_result = torch.sin(pi_tensor)
```

```python
# Plot the result
plt.plot(pi_tensor.numpy(),sin_result.numpy())
```

```python
# Construct a tensor with 25 steps in the range 0 to pi/2
tensor_pi = torch.linspace(0, np.pi/2 , 100)
max_tensor_pi = tensor_pi.max()
min_tensor_pi = tensor_pi.min()
print(max_tensor_pi)
print(min_tensor_pi)
sin_tensor_pi = torch.sin(tensor_pi)
plt.plot(tensor_pi.numpy(),sin_tensor_pi.numpy())
```

### Tensor Operations

### Addition

```python
# Create two sample tensor
u = torch.tensor([1,0])
v = torch.tensor([0,1])
```

```python
# Add u and v
w = u + v
print(w)
```

```python
# Add a scalar to the tensor
u = torch.tensor([1,2,3,-1])
v = u + 1
print(v)
```

### Tensor Multiplication

```python
u = torch.tensor([1,2])
v = u * 2
print(v)
```

```python
u = torch.tensor([1,2])
v = torch.tensor([3,2])
w = u * v
print(w)
```

### Dot product

```python
u = torch.tensor([1,2])
v = torch.tensor([3,2])
print(torch.dot(u,v))
# (1*3)+(2*2) = 7
```
